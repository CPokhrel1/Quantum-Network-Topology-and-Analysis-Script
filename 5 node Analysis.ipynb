```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Convert Picoseconds to MilliSeconds
df1["Start_time"] /= 1e12
df1["End_time"] /= 1e12
df1["Wait_time"] /= 1e12
```

```python
# Compute Latency
df1["Latency"] = df1["End_time"] - df1["Start_time"]
```

```python
# **Throughput Analysis**
peak_throughput = df1["Throughput"].max()
average_throughput = df1["Throughput"].mean()
throughput_variation = df1["Throughput"].std()
throughput_fidelity_corr = df1["Throughput"].corr(df1["Fidelity"])
```

```python
# **Latency Analysis**
average_latency = df1["Latency"].mean()
latency_fidelity_corr = df1["Latency"].corr(df1["Fidelity"])
latency_throughput_corr = df1["Latency"].corr(df1["Throughput"])
```

```python
# **Wait Time Analysis**
wait_time_fidelity_corr = df1["Wait_time"].corr(df1["Fidelity"])
wait_time_throughput_corr = df1["Wait_time"].corr(df1["Throughput"])
```

```python
# Print Analysis Results
print(f"Peak Entanglement Throughput: {peak_throughput:.2f} EBPS")
print(f"Average Entanglement Throughput: {average_throughput:.2f} EBPS")
print(f"Throughput Variation: {throughput_variation:.2f} EBPS")
print(f"Throughput-Fidelity Correlation: {throughput_fidelity_corr:.2f}")
print(f"Average Latency: {average_latency:.2f} milliseconds")
print(f"Latency-Fidelity Correlation: {latency_fidelity_corr:.2f}")
print(f"Latency-Throughput Correlation: {latency_throughput_corr:.2f}")
print(f"Wait Time-Fidelity Correlation: {wait_time_fidelity_corr:.2f}")
print(f"Wait Time-Throughput Correlation: {wait_time_throughput_corr:.2f}")
```
